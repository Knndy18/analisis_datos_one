### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



```python
import matplotlib.pyplot as plt
facturacion_tienda01 = sum(tienda['Precio'])
facturacion_tienda02 = sum(tienda2['Precio'])
facturacion_tienda03 = sum(tienda3['Precio'])
facturacion_tienda04 = sum(tienda4['Precio'])

facturacion_total = [facturacion_tienda01, facturacion_tienda02, facturacion_tienda03, facturacion_tienda04]

print(facturacion_total)


```

# 2. Ventas por categoría

```python
categoria_tienda01 =tienda.groupby('Categoría del Producto')['Precio'].sum()
categoria_tienda02 =tienda2.groupby('Categoría del Producto')['Precio'].sum()
categoria_tienda03 =tienda3.groupby('Categoría del Producto')['Precio'].sum()
categoria_tienda04 =tienda4.groupby('Categoría del Producto')['Precio'].sum()

```

# 3. Calificación promedio de la tienda


```python
calificacion_tienda01 = float(tienda['Calificación'].mean())
calificacion_tienda02 = float(tienda2['Calificación'].mean())
calificacion_tienda03 = float(tienda3['Calificación'].mean())
calificacion_tienda04 = float(tienda4['Calificación'].mean())

calificacion_total = [calificacion_tienda01, calificacion_tienda02, calificacion_tienda03, calificacion_tienda04]

print(calificacion_total)
```

# 4. Productos más y menos vendidos

```python
productos_vendidos_tienda01 = tienda["Producto"].value_counts()
producto_mas_vendido_tienda01 = productos_vendidos_tienda01.idxmax()
producto_menos_vendido_tienda01 = productos_vendidos_tienda01.idxmin()

print("Producto más vendido:", producto_mas_vendido_tienda01)
print("Producto menos vendido:", producto_menos_vendido_tienda01)
```

```python
productos_vendidos_tienda02 = tienda2["Producto"].value_counts()
producto_mas_vendido_tienda02 = productos_vendidos_tienda02.idxmax()
producto_menos_vendido_tienda02 = productos_vendidos_tienda02.idxmin()

print("Producto más vendido:", producto_mas_vendido_tienda02)
print("Producto menos vendido:", producto_menos_vendido_tienda02)

```

```python
productos_vendidos_tienda03 = tienda3["Producto"].value_counts()
producto_mas_vendido_tienda03 = productos_vendidos_tienda03.idxmax()
producto_menos_vendido_tienda03 = productos_vendidos_tienda03.idxmin()

print("Producto más vendido:", producto_mas_vendido_tienda03)
print("Producto menos vendido:", producto_menos_vendido_tienda03)
```

```python
productos_vendidos_tienda04 = tienda4["Producto"].value_counts()
producto_mas_vendido_tienda04 = productos_vendidos_tienda04.idxmax()
producto_menos_vendido_tienda04 = productos_vendidos_tienda04.idxmin()

print("Producto más vendido:", producto_mas_vendido_tienda04)
print("Producto menos vendido:", producto_menos_vendido_tienda04)
```

# 5. Envío promedio por tienda

```python
envio_promedio_tienda01 =float( tienda["Costo de envío"].mean())
envio_promedio_tienda02 =float( tienda2["Costo de envío"].mean())
envio_promedio_tienda03 =float( tienda3["Costo de envío"].mean())
envio_promedio_tienda04 =float( tienda4["Costo de envío"].mean())

envio_promedio_total= [envio_promedio_tienda01, envio_promedio_tienda02, envio_promedio_tienda03, envio_promedio_tienda04]

print(envio_promedio_total)
```

```python
print('grafico de calificacion')
#grafico calificacion
plt.plot(['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4'], calificacion_total,
         marker='o', linestyle='--', color='purple', linewidth=2)
plt.ylabel('Calificación promedio')
plt.title('Calificación promedio por tienda')
plt.grid(True, linestyle=':', alpha=0.5)
plt.ylim(0, 5)  # Establecer límites del eje Y si las calificaciones son de 0-5
plt.show()


print('grafico de facturacion')

#grafico facturacion
plt.bar(['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4'], facturacion_total)
plt.show()

print('grafico de envio')

#grafico envio
plt.plot(['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4'], envio_promedio_total,
         marker='o', linestyle='--', color='purple', linewidth=2)
plt.ylabel('Envío promedio')
```
